# Choosing the Right LLM for a Task

Today was a theory-focused session where we learned how to select the most suitable language model for a given problem rather than building something practical.

## Chinchilla Scaling Law

We studied the **Chinchilla Scaling Law**, which explains the relationship between a model's size and the amount of training data required. As the **number of parameters increases**, the model also requires **more training tokens** to be trained effectively. Simply increasing model size without providing enough training data can lead to inefficient performance.

As a general rule of thumb, **the number of parameters can be used as a rough indicator of a model's capability**, although it is not always the best or most accurate way to compare models.

## Common LLM Benchmarks

### GPQA (Google Proof Q&A)

A benchmark consisting of challenging Physics, Chemistry, and Biology questions at a PhD level. It contains **448 questions**. Interestingly, a well-informed person without a PhD, even with **30 minutes of internet access per question**, scores only about **34%**, making it a useful benchmark for evaluating advanced reasoning in LLMs.

### MMLU-Pro (Massive Multitask Language Understanding – Pro)

A comprehensive benchmark covering multiple academic disciplines using **multiple-choice questions with 10 answer options**, designed to measure broad knowledge and reasoning.

### AIME (American Invitational Mathematics Examination)

A benchmark based on highly challenging competitive mathematics problems that require deep mathematical reasoning and problem-solving skills.

### LiveCodeBench

A coding benchmark where models solve **previously unseen programming problems**, often sourced from platforms such as LeetCode contests. It evaluates real-world coding ability rather than memorization.

### MuSR (Multistep Soft Reasoning)

A reasoning benchmark in which the model is given a **1,000-word story**, often involving a murder mystery or other complex scenario, and must infer the correct solution through logical reasoning.

### HLE (Humanity's Last Exam)

One of the most difficult LLM benchmarks available, containing approximately **2,500 extremely challenging questions** spanning a wide range of subjects. It is designed to push models to their reasoning and knowledge limits.

## Model Leaderboards

Before selecting an LLM for a specific application, it is useful to **shortlist candidate models using public leaderboards**. These leaderboards evaluate models across numerous benchmarks and tasks, allowing us to compare their strengths in areas such as reasoning, coding, mathematics, and general knowledge.

We also learned **how these leaderboards are created**, including the variety of benchmark tests that models undergo before being ranked for public comparison.
