## LMArena

A practical way to compare LLMs is through **LMArena** (formerly Chatbot Arena), where users can interact with two anonymous models side by side and vote for the better response. This provides a more human-centric evaluation compared to benchmark scores alone.

**Website:** https://lmarena.ai

## Commercial LLM Application Types

### 1. GPT Wrapper

Applications that primarily **forward user requests to a frontier LLM** (such as GPT, Claude, or Gemini) and provide a chat interface with minimal additional functionality. These products mainly improve the user experience rather than the model itself.

### 2. Specialized AI Applications

Applications built for a **specific domain or business problem**. They typically combine frontier models with:
- Proprietary or domain-specific data
- Retrieval-Augmented Generation (RAG)
- Fine-tuning or custom prompts
- External tools and APIs

This allows the model to produce more accurate and context-aware responses for specialized tasks.

### 3. Agentic AI Systems

AI systems capable of **taking actions instead of only generating text**. Agentic systems can plan, reason, use tools, call APIs, browse the web, execute code, and complete multi-step tasks autonomously to achieve a goal.
